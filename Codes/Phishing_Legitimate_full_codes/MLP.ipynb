```python
# Importation des librairies nécessaires
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Prétraitement et Split
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_curve, auc

# Sélection de features
from sklearn.feature_selection import SelectKBest, f_classif

# Modèles deep learning avec Keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
import tensorflow as tf

# Détection d'anomalies
from sklearn.ensemble import IsolationForest

# Pour équilibrer la dataset avec SMOTE
from imblearn.over_sampling import SMOTE

# Pour le Bagging
import random

# ---------------------------
# 0. Chargement et exploration du dataset
# ---------------------------
df = pd.read_csv("Phishing_Legitimate_full.csv")

# Affichage d'un aperçu du dataset
print("Dimensions du dataset:", df.shape)
print("Aperçu des 5 premières lignes:")
print(df.head())

# Visualiser les informations générales du dataset
print("\nInformations sur le dataset:")
df.info()
print("\nStatistiques descriptives:")
print(df.describe())

# ---------------------------
# 1. Préparation des données
# ---------------------------
# Suppression des colonnes non pertinentes
if 'id' in df.columns:
    df.drop(columns=['id'], inplace=True)
if 'url' in df.columns:
    df.drop(columns=['url'], inplace=True)

# Conversion ou ré-encodage de la variable cible
# Ici nous supposons que la colonne cible est 'CLASS_LABEL' avec 0 (légitime) et 1 (phishing)
print("\nDistribution initiale des classes :", df['CLASS_LABEL'].value_counts())

# S'il existe des valeurs manquantes dans CLASS_LABEL, on les supprime
df.dropna(subset=['CLASS_LABEL'], inplace=True)

# Séparation des features et de la cible
X = df.drop(columns=['CLASS_LABEL']).values
y = df['CLASS_LABEL'].values

# ---------------------------
# 2. Normalisation des features
# ---------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ---------------------------
# 3. Suppression des anomalies avec IsolationForest
# ---------------------------
print("\nDétection d'anomalies pour différents niveaux de contamination :")
for contamination in [0.005, 0.01, 0.02]:
    iso = IsolationForest(contamination=contamination, random_state=42)
    pred_anomalies = iso.fit_predict(X_scaled)
    mask = pred_anomalies != -1
    nombre_supprimes = len(X_scaled) - np.sum(mask)
    print(f"Pour une contamination de {contamination}, points supprimés : {nombre_supprimes}")

# Choix du taux de contamination (ici 1%)
chosen_contamination = 0.01
iso = IsolationForest(contamination=chosen_contamination, random_state=42)
pred_anomalies = iso.fit_predict(X_scaled)
mask = pred_anomalies != -1  # On conserve les échantillons normaux
X_scaled = X_scaled[mask]
y = y[mask]

# ---------------------------
# 4. Équilibrage de la dataset avec SMOTE
# ---------------------------
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X_scaled, y)
print("\nDistribution après SMOTE :", np.bincount(y_res))

# ---------------------------
# 5. Sélection des meilleures features
# ---------------------------
# Utilisation de SelectKBest pour retenir les 20 features les plus significatives
selector = SelectKBest(score_func=f_classif, k=20)
selector.fit(X_res, y_res)
X_sel = selector.transform(X_res)
selected_feature_indices = selector.get_support(indices=True)
print("\nIndices des features sélectionnées :", selected_feature_indices)

# (Optionnel) Visualisation de l'importance relative des features via leurs scores
feature_scores = selector.scores_
all_features = df.drop(columns=['CLASS_LABEL']).columns
score_df = pd.DataFrame({'Feature': all_features, 'Score': feature_scores})
score_df.sort_values(by='Score', ascending=False, inplace=True)
plt.figure(figsize=(12,6))
sns.barplot(x="Score", y="Feature", data=score_df.head(20))
plt.title("Top 20 features selon f_classif")
plt.show()

# ---------------------------
# 6. Split train/test
# ---------------------------
X_train, X_test, y_train, y_test = train_test_split(X_sel, y_res,
                                                    test_size=0.2,
                                                    random_state=42,
                                                    stratify=y_res)

# ---------------------------
# 7. Définition du modèle MLP amélioré
# ---------------------------
def build_mlp(input_shape):
    model = Sequential([
        Dense(128, activation='relu', input_shape=(input_shape,)),
        BatchNormalization(),
        Dropout(0.3),
        Dense(64, activation='relu'),
        BatchNormalization(),
        Dropout(0.3),
        Dense(32, activation='relu'),
        BatchNormalization(),
        Dropout(0.3),
        Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam',
                  loss='binary_crossentropy',
                  metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()])
    return model

# ---------------------------
# 8. Fonction d'entraînement du Bagging
# ---------------------------
epochs = 100
batch_size = 64

callbacks = [
    EarlyStopping(patience=10, restore_best_weights=True),
    ModelCheckpoint('best_model.h5', save_best_only=True)
]

def train_bagging_model(n_estimators, X_train, y_train):
    bagging_models = []
    for i in range(n_estimators):
        print(f"\nEntraînement du modèle MLP {i+1}/{n_estimators}...")
        # Création d'un échantillon bootstrap
        indices = np.random.choice(np.arange(len(X_train)), size=len(X_train), replace=True)
        X_train_boot = X_train[indices]
        y_train_boot = y_train[indices]

        model = build_mlp(X_train.shape[1])
        model.fit(X_train_boot, y_train_boot,
                  epochs=epochs,
                  batch_size=batch_size,
                  validation_split=0.2,
                  callbacks=callbacks,
                  class_weight={0: 1, 1: 1.2},
                  verbose=0)
        bagging_models.append(model)
        print(f"Modèle {i+1} terminé.")
    return bagging_models

# ---------------------------
# 9. Expérimentation avec différents nombres d'estimateurs
# ---------------------------
for n in [3, 5, 7, 10]:
    print(f"\n=== Entraînement avec {n} estimateurs ===")
    bagging_models = train_bagging_model(n, X_train, y_train)

    # Agrégation des prédictions sur l'ensemble de test
    predictions = []
    for model in bagging_models:
        pred = model.predict(X_test)
        predictions.append(pred)

    # Calcul de la moyenne des prédictions
    predictions = np.array(predictions)  # forme: (n_estimators, n_samples, 1)
    avg_predictions = np.mean(predictions, axis=0)

    # Application du seuil 0.5 pour obtenir la classe finale
    y_pred = (avg_predictions >= 0.5).astype(int).ravel()

    # Évaluation des performances
    print("\n--- Rapport de classification ---")
    print(classification_report(y_test, y_pred))

    # Calcul et affichage de la courbe ROC
    proba = avg_predictions.ravel()
    fpr, tpr, thresholds = roc_curve(y_test, proba)
    roc_auc = auc(fpr, tpr)

    plt.figure(figsize=(8, 6))
    plt.plot(fpr, tpr, color='darkorange', lw=2, label='ROC (AUC = %0.2f)' % roc_auc)
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('Taux de faux positifs')
    plt.ylabel('Taux de vrais positifs')
    plt.title(f'Courbe ROC - Bagging (MLP) avec {n} estimateurs')
    plt.legend(loc="lower right")
    plt.show()
```
